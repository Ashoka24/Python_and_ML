Imagine you're helping a teacher grade a language test. In one of the tasks, students were asked 
to identify and list all the words from a passage that have a length equal to a prime number. 
Your task is to write a Python program that does this automatically.
The program will take a sentence or passage as input and output all the words whose length is a 
prime number.
Note: A prime number is a number greater than 1 that has no divisors other than 1 and itself.
Example 1:
Input:
The quick brown fox jumps over the lazy dog
Output:
The quick brown fox jumps the
Explanation:
Word Lengths "The" → 3 letters (prime) "quick" → 5 letters (prime) "brown" → 5 letters (prime) 
"fox" → 3 letters (prime) "jumps" → 5 letters (prime) "over" → 4 letters (not prime) "the" → 3 
letters (prime) "lazy" → 4 letters (not prime) "dog" → 3 letters (prime)

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2,int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
print(is_prime(2))
```

```python
def words_with_prime_length(s):
    words = s.split()
    prime_words = [word for word in words if is_prime(len(word))]
    return prime_words
print(words_with_prime_length("The quick brown fox jumps over the lazy dog"))
```

```python

def rearrange_list(nums):
    nums.sort()  # Step 1: Sort the list
    result = []  
    left, right = 0, len(nums) - 1  # Two pointers

    while left <= right:
        if left != right:
            result.append(nums[left])  # Pick the smallest
            result.append(nums[right])  # Pick the largest
        else:
            result.append(nums[left])  # Add the last middle element
        
        left += 1
        right -= 1

    return result
```

```python
nums = [1, 3, 4, 5, 11]
output = rearrange_list(nums)
print(output)
```

```python
#Write a function to validate an email address. A valid email has the format:

#📌 Pattern: username@domain.extension

#username → Contains letters, digits, _, . (but not at the start or end).
#domain → Contains letters and digits.
#extension → 2 to 4 letters (e.g., .com, .org, .edu)
import re
def validate_email(email):
    pattern = r"^[\w][a-zA-Z0-9._]*[\w]@[a-zA-Z0-9]+\.[a-zA-Z]{2,4}$"
    if re.match(pattern, email):
        return True
    return False
   

print(validate_email("username@domain.ext")) 
print(validate_email("user@domain"))
print(validate_email("user.@domain.com"))
```

```python
def is_valid_card(card):
    # Check proper format (either continuous 16 digits or in xxxx-xxxx-xxxx-xxxx format)
    pattern = re.compile(r'^(?:[4-6]\d{15}|[4-6]\d{3}(?:-\d{4}){3})$')
    if not pattern.match(card):
        return "Invalid"

    # Remove hyphens for repetition check
    card_num = card.replace('-', '')

    # Check for 4 or more consecutive repeated digits
    if re.search(r'(\d)\1{3,}', card_num):
        return "Invalid"

    return "Valid"


n = int(input())
for _ in range(n):
    card_number = input().strip()
    print(is_valid_card(card_number))
```
